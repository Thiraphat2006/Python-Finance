```python
from project.Untility.data_loader import DataLoader as dl
from project.strategy.Moving_Avg_Logic import MovingAvgLogic as ma_logic
from project.Indicator.Moving_Avg import MovingAvg
import vectorbt as vbt
from pprint import pprint as pp
from datetime import datetime
```

```python
symbol ="XAUUSD"
time_start = datetime(2020, 1, 1)
time_stop = datetime(2024, 1, 1)

data = dl(symbol).load_with_date(utc_from =time_start, utc_to = time_stop)

indicator_data = MovingAvg(data).compute_indicators()

result = ma_logic(indicator_data)()

pp(result.head()) 
pp(indicator_data.head())
```

```python
long_entries = result["long_entry"]
long_exits = result["exit_long"]
short_entries = result["short_entry"]
short_exits = result["exit_short"]

pf = vbt.Portfolio.from_signals(
    close=indicator_data["close"],
    entries = long_entries,
    exits = long_exits,
    short_entries = short_entries,
    short_exits = short_exits,
    init_cash = 10000,
    fees = 0.001,
    freq="15m"
)

```
